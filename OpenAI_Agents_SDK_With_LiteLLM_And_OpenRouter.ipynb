```python
from google.colab import userdata
OPENROUTER_API_KEY = userdata.get('OPENROUTER_API_KEY')
```

```python
base_url = "https://openrouter.ai/api/v1"
model_1 = "qwen/qwen3-235b-a22b:free"
```

# First Method With OpenRouter


```python
from agents import Agent, Runner, OpenAIChatCompletionsModel, AsyncOpenAI
from agents.run import RunConfig
import asyncio
```

```python
external_client = AsyncOpenAI(
    base_url=base_url,
    api_key=OPENROUTER_API_KEY
)

model = OpenAIChatCompletionsModel(
    model = model_1,
    openai_client=external_client
)

config = RunConfig(
    model = model,
    model_provider=external_client,
    tracing_disabled = True
)
```

```python
async def main():
  agent = Agent(
      name = "Qwen Greeting Agent",
      instructions="You are a greeting agent that greets the user with nice words",
      model = model
  )

  result = await Runner.run(
      agent,
      "Hello Assalamu Alaikum",
      run_config=config
  )

  print(result.final_output)
```

```python
asyncio.run(main())
```

# Second Method With OpenRouter

```python
from agents import Agent, Runner, OpenAIChatCompletionsModel, set_tracing_disabled
from openai import AsyncOpenAI
```

```python
set_tracing_disabled(disabled=True)

external_client = AsyncOpenAI(
    base_url=base_url,
    api_key=OPENROUTER_API_KEY
)

async def main_2():
  agent = Agent(
      name = "Mistrial Greeting Agent",
      instructions="You are a greeting agent that greets the users with nice words",
      model = OpenAIChatCompletionsModel(
          model = "mistralai/mistral-small-3.1-24b-instruct:free",
          openai_client=external_client
      ))
  result = await Runner.run(
      agent,
      "Hello,Good Morning"
  )
  print(result.final_output)
```

```python
asyncio.run(main_2())
```

```python
from agents import Agent, Runner, OpenAIChatCompletionsModel, set_tracing_disabled
from openai import AsyncOpenAI
from agents.extensions.models.litellm_model import LitellmModel
```

```python
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
```

```python
import asyncio

set_tracing_disabled(disabled=True)

MODEL = "gemini/gemini-2.0-flash"

async def main_3():

   agent = Agent(
     name = "Gemini Greeting Agent",
     instructions="You are a greeting agent that greets the users with nice words",
      model = LitellmModel(
         model = MODEL,
         api_key=GOOGLE_API_KEY,
         )
      )
   result = await Runner.run(
       agent,
      "Hello Good Morning"
   )
   print(result.final_output)
```
